```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure with a suitable size
plt.figure(figsize=(15, 9))

# ---------------- Plot 1 ----------------
xpoint = np.array([1, 8])
ypoint = np.array([3, 10])

plt.subplot(2, 3, 1)
plt.plot(xpoint, ypoint, marker='o', linestyle='dashed')
plt.title("Title 1")
plt.xlabel("X label", color="blue")
plt.ylabel("Y label", color="blue")


# ---------------- Plot 2 ----------------
y1 = np.array([3, 8, 1, 10])
y2 = np.array([6, 2, 7, 9])

plt.subplot(2, 3, 2)
plt.plot(y1, linewidth=2.5, color='orange', marker='*')
plt.plot(y2, linewidth=2.5, color='blue', linestyle='dashed')
plt.title("Two Line Plot")
plt.xlabel("X")
plt.ylabel("Y")


# ---------------- Plot 3 ----------------
ypoints = np.array([3, 8, 1, 10])

plt.subplot(2, 3, 3)
plt.plot(
    ypoints,
    marker='o',
    markersize=10,
    markeredgecolor='black',
    markerfacecolor='yellow'
)
plt.title("Customized Markers")
plt.xlabel("X")
plt.ylabel("Y")


# ---------------- Plot 4 ----------------
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.subplot(2, 3, 4)
plt.title("Title 2")
plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)


# ---------------- Plot 5 ----------------
x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])

sizes = np.array([42, 5, 12, 25, 14, 45, 20, 24, 16, 29, 11, 9, 30])
colors = np.array([0, 10, 20, 30, 40, 45, 50, 55, 60, 70, 80, 90, 100])

plt.subplot(2, 3, 5)
scatter = plt.scatter(
    x,
    y,
    c=colors,
    cmap='viridis',
    s=sizes,
    alpha=0.5
)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.colorbar(scatter, label="Color Value")


# ---------------- Plot 6 ----------------
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 3, 6)
plt.bar(x, y, color='red', width=0.5)
plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")


# Adjust layout
plt.tight_layout()

# Display all plots
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Pie chart
y = np.array([20, 35, 25, 20])

mylabels = ["20", "35", "25", "20"]
myexplode = [0, 0, 0.2, 0]
mycolors = ["red", "blue", "green", "yellow"]

plt.pie(
    y,
    labels=mylabels,
    startangle=90,
    explode=myexplode,
    shadow=True,
    colors=mycolors
)

plt.legend()
plt.show()
```
